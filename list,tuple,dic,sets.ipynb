```python
#slicing
num=[10,20,30,40,50,60,70,80]
print(num[-2:1:-3])
```

```python
num=[10,20,30,40,50,60,70,80]
print(num[1:6:2])
    
```

```python
#tuple
student=["manasa",18,"python"]
print(student[0])
print(student[1])
print(student[2])
```

```python
#count
num=[10,20,20,30,20]
print(num.count(20))
```

```python
#index
num=[10,20,30,40]
print(num.index(30))
```

```python
num=(10,20,30,40)
print(len(num))
print(max(num))
print(min(num))
print(sum(num))
```

```python
#sets
num={10,20,30,20,10}
print(num)
```

```python
#add values to a set 
sub={"python","java"}
sub.add("sql")
print(sub)
```

```python
#remove values from set
sub.remove("java")
print(sub)
```

```python
#sets do not allow duplicate values 
num={10,20,20,30,40,40,50}
print(num)
```

```python
#dictionaries in python
student={
    "name":"manasa",
    "age":18,
    "course":"cyber security"
}
print(student)
```

```python
#access elements in disc
print(student["name"])
print(student["age"])
print(student["course"])
```

```python
#add new data to dic
student["city"]="vijaywada"
print(student)
```

```python
#
print(student.keys())
print(student.values())
print(student.items())
print(student.get("name"))
student.update({"age":22})
print(student.pop("city"))
```

```python
#popitem
student.popitem()
print(student)
```

```python
student={
    "name":"manasa"
}
student.setdefault("age",21)
print(student)
```
