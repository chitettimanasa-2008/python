```python
#reverse a number
number=int(input("enter number:"))

reverse=0

while number>0:
    digit=number%10
    number=number//10
    reverse=reverse*10+digit

    print("reverse:",reverse)
```

```python
#check if a number is polindrome
number=int(input("enter a number:"))

original=number
reverse=0
while number>0:
    digit=number%10
    reverse=reverse*10+digit
    number=number//10
if original==reverse:
    print("polindrome")
else:
    print("not a polindrome")   
```
